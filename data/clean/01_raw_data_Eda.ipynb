```python
import pandas as pd
```

```python
file_path= "../raw/online_retail_II.xlsx"
excel_file = pd.ExcelFile(file_path)
excel_file.sheet_names
```

```python
for sheet in excel_file.sheet_names:
    sample = pd.read_excel(
        file_path,
        sheet_name=sheet,
        nrows=5
    )

    print(f"\nSHEET: {sheet}")
    print("Number of columns:", sample.shape[1])
    print("Columns:", sample.columns.tolist())
    display(sample)
```

```python
invoice_tables = []

for sheet in excel_file.sheet_names:
    invoice_df = pd.read_excel(
        file_path,
        sheet_name=sheet,
        usecols=["Invoice"]
    )

    invoice_tables.append(invoice_df)

    print(
        sheet,
        "unique invoices:",
        invoice_df["Invoice"].nunique()
    )
```
