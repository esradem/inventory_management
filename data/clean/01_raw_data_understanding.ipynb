## Data Cleaning

```python
#Prepare data for cleaning
df_clean = df_retail.copy()
```

### 1. Standardize column names 

```python
#Change column names lowcase, clean and replace spaces with underscores
df_clean.columns = ( 
    df_clean.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
print(df_clean.columns.tolist())
```

```python
### 2. Check/fix data types
df_clean.dtypes
df_clean['invoice'] = df_clean['invoice'].astype(str)
df_clean["customerid"] = df_clean["customerid"].astype("Int64")
df_clean["stockcode"] = df_clean["stockcode"].astype("string")
df_clean["description"] = df_clean["description"].astype("string")
```

```python
### 3. Investigate duplicates
duplicate_count = df_clean.duplicated().sum()

duplicate_count
```

```python
df_clean[df_clean.duplicated(keep=False)].head(20)
```

```python
df_clean = df_clean.drop_duplicates() 
df_clean.duplicated().sum()
```

### Duplicate handling:
 Exact duplicate transaction rows were found. Since identical records would otherwise double-count quantities and revenue, exact duplicates were removed. As the dataset does not contain a unique line-item identifier, this is treated as a documented cleaning assumption.

```python
### To make a RFM analysis we keep not na customerids in a copyied data frame
df_customer = df_clean[
    df_clean["customerid"].notna()
].copy()
df_customer["customerid"].nunique()
```

### Which transactions should count as a real completed purchase for RFM?
--- Recency (R): How recently a customer made a purchase. Recent buyers are more engaged and likely to buy again.
--- Frequency (F): How often a customer makes purchases within a set period. Higher frequency shows loyalty and habit.
--- Monetary Value (M): How much money a customer spends on purchases overall. This identifies your biggest financial contributors.

That's where we look at Invoice starting with C and negative Quantity.
For RFM, we want completed purchase transactions only. So we’ll create a separate DataFrame and exclude cancelled/invalid sales from that one.

```python
df_rfm = df_customer.copy()
print('Number of cancelled orders:', (df_rfm["invoice"].astype(str).str.startswith("C")).sum())
print('Number of negative quantity items:', (df_rfm["quantity"] <= 0).sum())
print('Number of negative prices:', (df_rfm["price"] < 0).sum())
```

```python
df_rfm = df_rfm[
    (~df_rfm["invoice"].astype(str).str.startswith("C")) &
    (df_rfm["quantity"] > 0) &
    (df_rfm["price"] > 0)
].copy()
```

### RFM analysis 1 step is look at frequency 

Frequency

How many unique orders did each customer make?


```python
frequency = (
    df_rfm.groupby("customerid")["invoice"]
    .nunique()
    .rename("frequency")
)
frequency.sort_values(ascending=False).head(20)
```

### Second step total revenue for each costumer

```python
# calculate the revenue to see how much each customer spend.
df_rfm['revenue'] = df_rfm['price'] * df_rfm['quantity']
```

```python
#monetary shows customer and revanue in total
monetary = (
    df_rfm.groupby('customerid')['revenue'].sum()
    .rename('monetary')
)
```

```python
# highesst spending customers
monetary.sort_values(ascending=False).head(10)
```

```python
df_rfm.columns.to_list()
```

```python
#Calculate Recency
#Find each customer’s latest purchase date, then calculate how many days ago that was relative to the dataset end date.
latest_purchase = df_rfm.groupby("customerid")["invoicedate"].max()
reference_date = df_rfm["invoicedate"].max() + pd.Timedelta(days=1)
```

```python
recency = (reference_date - latest_purchase).dt.days
recency = recency.rename("ecency")
```

```python
#look at our dataset rfm
rfm = pd.concat(
    [recency, frequency, monetary],
    axis=1
)
```
