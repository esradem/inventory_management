```python
import pandas as pd
```

```python
#read the excel file
file_path= "../raw/online_retail_II.xlsx"
excel_file = pd.ExcelFile(file_path)
excel_file.sheet_names
```

```python
df_2009 = pd.read_excel(file_path, sheet_name='Year 2009-2010')
df_2010 = pd.read_excel(file_path, sheet_name='Year 2010-2011')
```

```python
### How many rows?
print(df_2009.shape)
print(df_2010.shape)

###check if the columns match
df_2009.columns.equals(df_2010.columns)

##Merge the two dataframes into one dataframe
df_retail = pd.concat(
    [df_2009, df_2010],
    ignore_index=True
)
```

```python
### What are the columns? data types and total number of rows?
df_retail.info()
```

```python
#Change column name 
df_retail.rename(columns={"Customer ID": "CustomerID"}, inplace=True)
```

```python
df_retail.columns.tolist()
```

```python
# How many unique invoices?
print(f"Unique invoices: {df_retail['Invoice'].nunique()}")
```

```python
# How many unique products?
print(f"Unique product stockcodes: {df_retail['StockCode'].nunique()}")
print(f"Unique product descriptions: {df_retail['Description'].nunique()}")
```

```python
### How many unique customers?
print(f"Unique customers: {df_retail['CustomerID'].nunique()}")
```

```python
df_retail.groupby("CustomerID")["Invoice"].nunique().sort_values(ascending=False).head(10)
```

```python
### How much missing data?
missing_values = df_retail.isna().sum()
missing_percentage = df_retail.isna().mean()*100

print(missing_values)
print(missing_percentage)
```

```python
### How many cancellations?
cancelled = df_retail[
    df_retail["Invoice"].astype(str).str.startswith("C")
]
print(f"Number of cancelled invoices: {cancelled.shape[0]}")
print(f"Number of unique cancelled invoice IDs: {cancelled['Invoice'].nunique()}")
```
